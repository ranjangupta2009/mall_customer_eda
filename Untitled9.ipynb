```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# ---------------------------------------------------------
# 1. Generate 200-Row Customer Dataset (No CSV File Needed!)
# ---------------------------------------------------------
np.random.seed(42)
n_rows = 200

data = {
    'CustomerID': np.arange(1, n_rows + 1),
    'Gender': np.random.choice(['Female', 'Male'], size=n_rows, p=[0.56, 0.44]),
    'Age': np.random.randint(18, 71, size=n_rows),
    'Annual Income (k$)': np.random.randint(15, 138, size=n_rows),
    'Spending Score (1-100)': np.random.randint(1, 100, size=n_rows)
}

df = pd.DataFrame(data)

print("--- First 5 Rows (head) ---")
print(df.head(), "\n")

print("--- Last 5 Rows (tail) ---")
print(df.tail(), "\n")

print("--- Dataset Shape ---")
print(f"Rows: {df.shape[0]}, Columns: {df.shape[1]}\n")

print("--- Dataset Info ---")
df.info()
print("\n")

print("--- Statistical Summary (describe) ---")
print(df.describe(), "\n")

# ---------------------------------------------------------
# Question (6): Check for missing values and duplicates
# ---------------------------------------------------------
print("--- Missing Values per Column ---")
print(df.isnull().sum(), "\n")

print(f"Total Duplicate Rows: {df.duplicated().sum()}\n")

# ---------------------------------------------------------
# Question (1): Count customers by gender
# ---------------------------------------------------------
print("--- (1) Customer Count by Gender ---")
gender_counts = df['Gender'].value_counts()
print(gender_counts, "\n")

# ---------------------------------------------------------
# Question (2): Mean, median, min, and max using NumPy
# ---------------------------------------------------------
print("--- (2) Numerical Statistics (Using NumPy Arrays) ---")
numeric_cols = ['Age', 'Annual Income (k$)', 'Spending Score (1-100)']

for col in numeric_cols:
    arr = df[col].to_numpy()
    print(f"{col}:")
    print(f"  Mean:   {np.mean(arr):.2f}")
    print(f"  Median: {np.median(arr):.2f}")
    print(f"  Min:    {np.min(arr)}")
    print(f"  Max:    {np.max(arr)}\n")

# ---------------------------------------------------------
# Question (3): Which age group has the highest average spending score?
# ---------------------------------------------------------
print("--- (3) Average Spending Score by Age Group ---")
bins = [17, 25, 35, 45, 55, 75]
labels = ['18-25', '26-35', '36-45', '46-55', '56+']
df['Age_Group'] = pd.cut(df['Age'], bins=bins, labels=labels)

age_group_spending = df.groupby('Age_Group', observed=False)['Spending Score (1-100)'].mean().sort_values(ascending=False)
print(age_group_spending, "\n")
print(f"Highest spending age group: {age_group_spending.index[0]} (Avg Score: {age_group_spending.iloc[0]:.2f})\n")

# ---------------------------------------------------------
# Question (4): Compare average income and spending score by gender
# ---------------------------------------------------------
print("--- (4) Average Income & Spending Score by Gender ---")
gender_comparison = df.groupby('Gender')[['Annual Income (k$)', 'Spending Score (1-100)']].mean()
print(gender_comparison, "\n")

# ---------------------------------------------------------
# Question (5): List the 10 customers with the highest spending scores
# ---------------------------------------------------------
print("--- (5) Top 10 Customers by Spending Score ---")
top_10_spenders = df.sort_values(by='Spending Score (1-100)', ascending=False).head(10)
print(top_10_spenders[['CustomerID', 'Gender', 'Age', 'Annual Income (k$)', 'Spending Score (1-100)']], "\n")

# Filtering Example
high_income_high_spend = df[(df['Annual Income (k$)'] > 80) & (df['Spending Score (1-100)'] > 80)]
print(f"Number of high-income (>80k$) & high-spending (>80) customers: {len(high_income_high_spend)}\n")

# ---------------------------------------------------------
# Charts: Gender-count bar chart & Income vs Spending scatter plot
# ---------------------------------------------------------
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Chart 1: Gender-Count Bar Chart
axes[0].bar(gender_counts.index, gender_counts.values, color=['#ff9999', '#66b3ff'], edgecolor='black')
axes[0].set_title('Customer Count by Gender')
axes[0].set_xlabel('Gender')
axes[0].set_ylabel('Number of Customers')

# Chart 2: Scatter Plot of Annual Income vs Spending Score
axes[1].scatter(df['Annual Income (k$)'], df['Spending Score (1-100)'], color='teal', alpha=0.7)
axes[1].set_title('Annual Income vs. Spending Score')
axes[1].set_xlabel('Annual Income (k$)')
axes[1].set_ylabel('Spending Score (1-100)')
axes[1].grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```
